```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")

matches = pd.read_csv(
    "../data/processed/matches_cleaned.csv"
)

deliveries = pd.read_csv(
    "../data/processed/deliveries_cleaned.csv"
)

player_stats = pd.read_csv(
    "../data/processed/player_stats.csv"
)

bowling_stats = pd.read_csv(
    "../data/processed/bowling_stats.csv"
)

team_stats = pd.read_csv(
    "../data/processed/team_stats.csv"
)

season_stats = pd.read_csv(
    "../data/processed/season_stats.csv"
)

print("All datasets loaded successfully!")

print("Matches:", matches.shape)
print("Deliveries:", deliveries.shape)
print("Player stats:", player_stats.shape)
print("Bowling stats:", bowling_stats.shape)
print("Team stats:", team_stats.shape)
print("Season stats:", season_stats.shape)
print(matches.head())

matches_per_season = (
    matches
    .groupby("season")
    .size()
    .reset_index(name="matches")
)

print(matches_per_season)
plt.figure(figsize=(12, 6))

plt.plot(
    matches_per_season["season"],
    matches_per_season["matches"],
    marker="o"
)

plt.title("IPL Matches Played by Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

season_runs = (
     matches
    .groupby("season")["total_runs"]
    .sum()
    .reset_index()
)

print(season_runs)
plt.figure(figsize=(12, 6))

plt.plot(
    season_runs["season"],
    season_runs["total_runs"],
    marker="o"
)

plt.title("Total IPL Runs by Season")
plt.xlabel("Season")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

avg_score = (
    matches
    .groupby("season")["first_innings_score"]
    .mean()
    .reset_index()
)

avg_score["first_innings_score"] = (
    avg_score["first_innings_score"].round(2)
)

print(avg_score)
plt.figure(figsize=(12, 6))

plt.plot(
    avg_score["season"],
    avg_score["first_innings_score"],
    marker="o"
)

plt.title("Average First-Innings Score by Season")
plt.xlabel("Season")
plt.ylabel("Average Score")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

team_wins = (
    matches["winner"]
    .value_counts()
    .reset_index()
)

team_wins.columns = [
    "team",
    "wins"
]

print(team_wins.head(10))
plt.figure(figsize=(12, 6))

sns.barplot(
    data=team_wins.head(10),
    x="wins",
    y="team"
)

plt.title("Top IPL Teams by Number of Wins")
plt.xlabel("Wins")
plt.ylabel("Team")

plt.tight_layout()
plt.show()

qualified_teams = team_stats[
    team_stats["matches_played"] >= 50
].sort_values(
    "win_percentage",
    ascending=False
)

print(qualified_teams)
plt.figure(figsize=(12, 6))

sns.barplot(
    data=qualified_teams,
    x="win_percentage",
    y="team"
)

plt.title(
    "Team Win Percentage "
    "(Minimum 50 Matches)"
)

plt.xlabel("Win Percentage")
plt.ylabel("Team")

plt.tight_layout()
plt.show()

top_run_scorers = (
    player_stats
    .sort_values("runs", ascending=False)
    .head(10)
)

print(
    top_run_scorers[
        ["batter", "runs", "balls", "strike_rate"]
    ]
)
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_run_scorers,
    x="runs",
    y="batter"
)

plt.title("Top 10 IPL Run Scorers")
plt.xlabel("Runs")
plt.ylabel("Player")

plt.tight_layout()
plt.show()

top_six_hitters = (
    player_stats
    .sort_values("sixes", ascending=False)
    .head(10)
)

print(
    top_six_hitters[
        ["batter", "sixes", "runs"]
    ]
)
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_six_hitters,
    x="sixes",
    y="batter"
)

plt.title("Top 10 Six Hitters")
plt.xlabel("Sixes")
plt.ylabel("Player")

plt.tight_layout()
plt.show()

top_four_hitters = (
    player_stats
    .sort_values("fours", ascending=False)
    .head(10)
)

print(top_four_hitters[
    ["batter", "fours", "runs"]
])
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_four_hitters,
    x="fours",
    y="batter"
)

plt.title("Top 10 Four Hitters")
plt.xlabel("Fours")
plt.ylabel("Player")

plt.tight_layout()
plt.show()

qualified_batters = player_stats[
    player_stats["balls"] >= 500
]
top_strike_rate = (
    qualified_batters
    .sort_values(
        "strike_rate",
        ascending=False
    )
    .head(10)
)

print(
    top_strike_rate[
        ["batter", "runs", "balls", "strike_rate"]
    ]
)
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_strike_rate,
    x="strike_rate",
    y="batter"
)

plt.title(
    "Top Strike Rates "
    "(Minimum 500 Balls)"
)

plt.xlabel("Strike Rate")
plt.ylabel("Player")

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=qualified_batters,
    x="runs",
    y="strike_rate",
    size="sixes",
    alpha=0.7
)

plt.title(
    "Batting Performance: Runs vs Strike Rate"
)

plt.xlabel("Total Runs")
plt.ylabel("Strike Rate")

plt.tight_layout()
plt.show()

top_wicket_takers = (
    bowling_stats
    .sort_values(
        "wickets",
        ascending=False
    )
    .head(10)
)

print(
    top_wicket_takers[
        ["bowler", "wickets", "economy"]
    ]
)
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_wicket_takers,
    x="wickets",
    y="bowler"
)

plt.title("Top 10 IPL Wicket Takers")
plt.xlabel("Wickets")
plt.ylabel("Bowler")

plt.tight_layout()
plt.show()

qualified_bowlers = bowling_stats[
    bowling_stats["balls"] >= 600
]
best_economy = (
    qualified_bowlers
    .sort_values("economy")
    .head(10)
)

print(
    best_economy[
        ["bowler", "wickets", "economy"]
    ]
)
plt.figure(figsize=(12, 6))

sns.barplot(
    data=best_economy,
    x="economy",
    y="bowler"
)

plt.title(
    "Lowest Economy Rates "
    "(Minimum 600 Balls)"
)

plt.xlabel("Economy Rate")
plt.ylabel("Bowler")

plt.tight_layout()
plt.show()

matches["toss_winner_won"] = (
    matches["toss_winner"] ==
    matches["winner"]
)
toss_impact = (
    matches["toss_winner_won"]
    .value_counts(normalize=True)
    * 100
)

print(toss_impact)
percentage = (
    matches["toss_winner_won"].mean() * 100
)

print(
    f"Toss winner also won "
    f"{percentage:.2f}% of matches."
)

first_innings_team = (
    deliveries[
        deliveries["inning"] == 1
    ]
    .groupby("match_id")["batting_team"]
    .first()
    .reset_index()
)

first_innings_team.columns = [
    "id",
    "first_batting_team"
]
matches = matches.merge(
    first_innings_team,
    on="id",
    how="left"
)
matches["batting_first_won"] = (
    matches["winner"] ==
    matches["first_batting_team"]
)
batting_first_win_rate = (
    matches["batting_first_won"]
    .mean() * 100
)

print(
    "Batting-first win rate:",
    round(batting_first_win_rate, 2),
    "%"
)
chasing_win_rate = 100 - batting_first_win_rate

print(
    "Chasing win rate:",
    round(chasing_win_rate, 2),
    "%"
)

powerplay = deliveries[
    deliveries["over"] < 6
]
powerplay_team_match = (
    powerplay
    .groupby(
        ["match_id", "batting_team"]
    )["total_runs"]
    .sum()
    .reset_index()
)
powerplay_team_avg = (
    powerplay_team_match
    .groupby("batting_team")["total_runs"]
    .mean()
    .sort_values(ascending=False)
)

print(powerplay_team_avg)
plt.figure(figsize=(12, 7))

powerplay_team_avg.head(10).plot(
    kind="bar"
)

plt.title(
    "Average Powerplay Runs by Team"
)

plt.xlabel("Team")
plt.ylabel("Average Powerplay Runs")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

death_overs = deliveries[
    deliveries["over"] >= 16
]
death_team_match = (
    death_overs
    .groupby(
        ["match_id", "batting_team"]
    )["total_runs"]
    .sum()
    .reset_index()
)
death_team_avg = (
    death_team_match
    .groupby("batting_team")["total_runs"]
    .mean()
    .sort_values(ascending=False)
)
plt.figure(figsize=(12, 7))

death_team_avg.head(10).plot(
    kind="bar"
)

plt.title(
    "Average Death-Over Runs by Team"
)

plt.xlabel("Team")
plt.ylabel("Average Death-Over Runs")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(death_team_avg)

venue_matches = (
    matches["venue"]
    .value_counts()
    .head(15)
)

print(venue_matches)
plt.figure(figsize=(12, 7))

venue_matches.sort_values().plot(
    kind="barh"
)

plt.title("Top IPL Venues by Matches Hosted")
plt.xlabel("Matches")
plt.ylabel("Venue")

plt.tight_layout()
plt.show()

venue_score = (
    matches
    .groupby("venue")["first_innings_score"]
    .agg(
        matches="count",
        avg_score="mean"
    )
    .reset_index()
)
venue_score_filtered = venue_score[
    venue_score["matches"] >= 20
].sort_values(
    "avg_score",
    ascending=False
)
print(
    venue_score_filtered.head(10)
)

player_match_runs = (
    deliveries
    .groupby(
        ["match_id", "batter"]
    )["batsman_runs"]
    .sum()
    .reset_index()
)
player_consistency = (
    player_match_runs
    .groupby("batter")["batsman_runs"]
    .agg(
        matches="count",
        avg_runs="mean",
        std_runs="std"
    )
    .reset_index()
)
consistent_players = player_consistency[
    player_consistency["matches"] >= 30
].sort_values(
    "std_runs"
)

print(consistent_players.head(10))

# 1. Define target columns
correlation_columns = [
    "total_match_runs",
    "first_innings_score",
    "second_innings_score",
    "sixes",
    "fours",
    "wickets"
]

# 2. Filter to only columns present in matches
existing_cols = [col for col in correlation_columns if col in matches.columns]

# 3. Calculate and print correlation matrix safely
if existing_cols:
    correlation_matrix = matches[existing_cols].select_dtypes(include="number").corr()
    print(correlation_matrix)
else:
    print("None of the specified correlation columns were found in matches DataFrame.")
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("IPL Match Metrics Correlation")

plt.tight_layout()
plt.show()

eda_summary = pd.DataFrame({
    "Analysis": [
        "Total Matches",
        "Total Runs",
        "Total Sixes",
        "Total Fours",
        "Toss Winner Match Win %",
        "Batting First Win %",
        "Chasing Win %"
    ],
    
    "Value": [
        len(matches),
        deliveries["total_runs"].sum(),
        (deliveries["batsman_runs"] == 6).sum(),
        (deliveries["batsman_runs"] == 4).sum(),
        round(percentage, 2),
        round(batting_first_win_rate, 2),
        round(chasing_win_rate, 2)
    ]
})

print(eda_summary)


# 1. Create the images directory if it doesn't exist
os.makedirs("../images", exist_ok=True)
os.makedirs("../data/processed", exist_ok=True)

# 2. Save CSV summary
eda_summary.to_csv("../data/processed/eda_summary.csv", index=False)
print("CSV saved to:", os.path.abspath("../data/processed/eda_summary.csv"))

# 3. Save plot image BEFORE calling plt.show()
plt.savefig("../images/top_run_scorers.png", dpi=300, bbox_inches="tight")
print("Image saved to:", os.path.abspath("../images/top_run_scorers.png"))

# 4. Display plot in notebook
plt.show()


os.makedirs("../images", exist_ok=True)

def save_chart(filename):
    filepath = f"../images/{filename}"
    plt.savefig(filepath, dpi=300, bbox_inches="tight")
    print(f"Saved: {filepath}")
    plt.show()

# 1. matches_by_season.png
matches["season"].value_counts().sort_index().plot(kind="bar", color="skyblue")
plt.title("Matches Per Season")
plt.xlabel("Season")
plt.ylabel("Match Count")
save_chart("matches_by_season.png")

# 2. runs_by_season.png
deliveries.groupby("match_id")["total_runs"].sum().reset_index() \
    .merge(matches[["id", "season"]], left_on="match_id", right_on="id") \
    .groupby("season")["total_runs"].sum().plot(kind="line", marker="o", color="green")
plt.title("Total Runs Per Season")
save_chart("runs_by_season.png")

# 3. average_score_by_season.png
deliveries.groupby(["match_id", "inning"])["total_runs"].sum().reset_index() \
    .merge(matches[["id", "season"]], left_on="match_id", right_on="id") \
    .groupby("season")["total_runs"].mean().plot(kind="bar", color="orange")
plt.title("Average Score Per Season")
save_chart("average_score_by_season.png")

# 4. team_wins.png
matches["winner"].value_counts().head(10).plot(kind="barh", color="purple")
plt.title("Top Winning Teams")
save_chart("team_wins.png")

# 5. top_run_scorers.png
deliveries.groupby("batter")["batsman_runs"].sum().nlargest(10).plot(kind="bar", color="navy")
plt.title("Top Run Scorers")
save_chart("top_run_scorers.png")

# 6. top_wicket_takers.png
deliveries[deliveries["dismissal_kind"].notna() & (~deliveries["dismissal_kind"].isin(["run out", "retired hurt"]))] \
    .groupby("bowler").size().nlargest(10).plot(kind="bar", color="crimson")
plt.title("Top Wicket Takers")
save_chart("top_wicket_takers.png")

# 7. strike_rate_analysis.png
batsman_stats = deliveries.groupby("batter").agg(runs=("batsman_runs", "sum"), balls=("ball", "count"))
batsman_stats["strike_rate"] = (batsman_stats["runs"] / batsman_stats["balls"]) * 100
batsman_stats[batsman_stats["balls"] >= 500]["strike_rate"].nlargest(10).plot(kind="bar", color="teal")
plt.title("Top Strike Rates (Min 500 Balls)")
save_chart("strike_rate_analysis.png")

# 8. toss_analysis.png
matches["toss_decision"].value_counts().plot(kind="pie", autopct="%1.1f%%", colors=["gold", "lightgreen"])
plt.title("Toss Decision Distribution")
plt.ylabel("")
save_chart("toss_analysis.png")

# 9. powerplay_analysis.png
deliveries[deliveries["over"] <= 6].groupby("match_id")["total_runs"].sum().plot(kind="hist", bins=20, color="coral")
plt.title("Powerplay Runs Distribution (Overs 1-6)")
save_chart("powerplay_analysis.png")

# 10. death_over_analysis.png
deliveries[deliveries["over"] >= 16].groupby("match_id")["total_runs"].sum().plot(kind="hist", bins=20, color="darkred")
plt.title("Death Over Runs Distribution (Overs 16-20)")
save_chart("death_over_analysis.png")

# 11. venue_analysis.png
matches["venue"].value_counts().head(10).plot(kind="barh", color="darkgreen")
plt.title("Top 10 Venues Hosted")
save_chart("venue_analysis.png")    
```
